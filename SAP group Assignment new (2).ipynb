```python
name=input("your name:")
weight=eval(input("weight(Kg):"))
height=eval(input("height(cm):"))

planet=input("what planet are you in (E/M):").upper()
if planet=="E":
  actual_weight=weight
  actual_height=height
elif planet=="M":
  actual_weight=weight/0.38
  actual_height=height-5
else:
  print("error")
print(f"actual weight: {actual_weight:.2f}")
print(f"actual height: {actual_height:.2f}")

calorie_lost=0
bmi=actual_weight/(actual_height/100)**2
answer="Y"

print(f"BMI: {bmi:.2f}")
if bmi>=40:
  status="obese"
elif bmi>=25:
  status="overweight"
elif bmi>=18.5:
  status="normal"
else:
  status="underweight"
print(f"status: {status}")


goal=eval(input("weight loss goal(kg):"))
calorie_goal=goal*7700
print(f"calorie loss goal:{calorie_goal:.2f}")

while answer!="N":
  workout=input("what workout you want (P/R):").upper()

  if workout=="P":
    print("push up")
    level=input("level (E/N/H):").upper()
    if level=="E":
      push_up=10
    elif level=="N":
      push_up=20
    elif level=="H":
      push_up=40

    print(f"Push up: {push_up}")
    if planet=="E":
      calorie_loss=push_up*0.5
    elif planet=="M":
      calorie_loss=push_up*0.19

  calorie_lost += calorie_loss


  if workout=="R":
    print("Running")
    level=input("level (E/N/H):").upper()
    if level=="E":
      run=5
    elif level=="N":
      run=10
    elif level=="H":
      run=21

    print(f"Running(km): {run}")
    if planet=="E":
      calorie_loss=run*weight
    elif planet=="M":
      calorie_loss=run*weight*1.15

  calorie_lost += calorie_loss

  if calorie_lost >= calorie_goal:
    print("you have reach your goal!")
    print("Ready for REBIRTH!")
    answer="N"
  elif calorie_lost < calorie_goal:
    print("you have not reach your goal!")
    answer=input("do you want to continue (Y/N):").upper()


weight_loss=calorie_lost/7700
current_weight=weight-weight_loss
level=calorie_lost/calorie_goal*100
print("==== FINAL RESULT ====")
print(f"weight loss: {weight_loss:.2f}")
print(f"current weight: {current_weight:.2f}")
print(f"LEVEL: {level:.1f}")
```
