```python
print("=" * 40)
print("       🚀 LEVEL UP 2050 🚀")
print("        SPACE FITNESS APP")
print("           CREW PROFILE")
print("=" * 40)
name=input("👨‍🚀 Enter your name:  ")
#Personal information
weight=eval(input("⚖️ Weight(KG)     :  "))
height=eval(input("📏 Height(CM)     :  "))

print("\n🌍 PLANET SELECTION")
print("[E] 🌎 EARTH")
print("[M] 🔴 MARS")
planet=input("What planet are you on (E/M): ").upper()
print("-" * 40)
print("          HEALTH INFORMATION")
print("-" * 40)
if planet=="E":
  print("🌍 Planet       : EARTH")
  actual_weight=weight
  actual_height=height
elif planet=="M":
  print("🔴 Planet       : MARS")
  actual_weight=weight/0.38
  actual_height=height-5
else:
  print("Invalid planet! Please enter E or M")
  exit()


print(f"⚖️ Actual weight: {actual_weight:.2f}")
print(f"📏 Actual height: {actual_height:.2f}")

calorie_lost=0

#BMI Calculation
bmi=actual_weight/(actual_height/100)**2
answer="Y"

print(f"📊 BMI          : {bmi:.2f}")
if bmi>=40:
  status="OBESE"
elif bmi>=25:
  status="OVERWEIGHT"
elif bmi>=18.5:
  status="NORMAL"
else:
  status="UNDERWEIGHT"
print(f"💚 Status       : {status}")

print("\n" + "=" * 40)
print("          🎯 MISSION TARGET")
print("=" * 40)
goal=eval(input("Enter weight loss goal(kg):  "))
calorie_goal=goal*7700
print(f"Calorie loss goal:  {calorie_goal:.2f}")
print("-" * 40)
print("YOUR MISSION BEGINS NOW!")

while answer!="N":
  print("\n" + "=" * 40)
  print("          🏋️ WORKOUT MENU")
  print("=" * 40)
  print("[P] 💪 PUSH UP")
  print("[R] 🏃 RUNNING")
  print()
  workout=input("Choose Workout (P/R):  ").upper()
  print("\n" + "=" * 40)

  if workout=="P":
    print("[]"*20)
    print("\n💪 PUSH UP")
    level=input("level (E/N/H): ").upper()
    if level=="E":
      push_up=10
    elif level=="N":
      push_up=20
    elif level=="H":
      push_up=40

    print(f"Push up _>_> {push_up}")
    if planet=="E":
      calorie_loss=push_up*0.5
    elif planet=="M":
      calorie_loss=push_up*0.19

  calorie_lost += calorie_loss


  if workout=="R":
    print("~!"*20)
    print("\n🏃 RUNNING")
    level=input("level (E/N/H): ").upper()
    if level=="E":
      run=5
    elif level=="N":
      run=10
    elif level=="H":
      run=21

    print(f"Running Distance === {run} KM")
    if planet=="E":
      calorie_loss=run*weight
    elif planet=="M":
      calorie_loss=run*weight*1.15

  calorie_lost += calorie_loss

  if calorie_lost >= calorie_goal:
    print("\n🏆 MISSION COMPLETED!")
    print("READY FOR LEVEL UP!")
    answer="N"
  elif calorie_lost < calorie_goal:
    print("\n🚀 MISSION IN PROGRESS!")
    answer = input("Continue workout? (Y/N): ").upper()


weight_loss=calorie_lost/7700
current_weight=weight-weight_loss
level=calorie_lost/calorie_goal*100
print("\n" + "=" * 40)
print("          🔥 MISSION STATUS")
print("=" * 40)

print(f"⚖️ Starting weight : {weight:.2f} kg")
print(f"📉 Weight lost     : {weight_loss:.2f} kg")
print(f"⚖️ Current weight  : {current_weight:.2f} kg")
print(f"🏁 LEVEL           : {level:.1f}")
print("-" * 45)
```
